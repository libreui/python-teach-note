```python
# /除法 // 整除
a = 10
b = 2
print("除法：", a / b, type(a / b))
print("整除：", a // b, type(a // b))
```

```python
# 指数 **
a = 3 ** 4
print("指数：", a, type(a))
```

```python
# 取余 %
a = 10
b = 3

# a 除 b 的余数
print("取余：", a % b, type(a % b))

```

```python
"""
算术运算符
- 加法 `+` "和是："
- 减法 `-` "差是："
- 乘法 `*` "积是："
- 除法 `/` （返回浮点数float）
- 整除 `//` （返回商的整数部分int）
- 取余 `%` "余数是："
- 幂 `**` "幂是："
"""

a, b = 10, 3
print(f"和是：{a + b}")
print(0.9 + 0.1)
```

```python
a = 3432
print(f"个位：{a % 10}")
print(f"十位：{a // 10 % 10}")
print(f"百位: {a // 100 % 10}")
print(f"千位：{a // 1000}")
x1 = a % 10
x2 = a // 10 % 10
x3 = a // 100 % 10
x4 = a // 1000
c = str(x1) + str(x2)
print(int(c), x1, x2, x3, x4)
```
