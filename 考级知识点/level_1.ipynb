```python
# is
# 变量 = 值(类型：数，字符串，布尔)
a = [2, 3]
b = [2, 3]
#
# bool布尔值 True, False
print(a == b)
print(a is b)
print(a is not b)

# 1 and 2 and 0

a = 4
a = a ** 2
print(a)
```
